```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Chargement du fichier des champignons
df = pd.read_csv('DataTP1_7de8ef50.csv')

# 2. Analyse Exploratoire des Données (AED)
print("Forme du dataset :", df.shape)
print("Colonnes disponibles :", df.columns.tolist())
display(df.head())
display(df.describe(include="all"))

# 3. Traitement des valeurs manquantes
print("Valeurs manquantes par colonne :")
print(df.isnull().sum())

# Exemple de traitement : Remplir les valeurs manquantes de 'cap-surface' par la valeur la plus fréquente (mode)
if 'cap-surface' in df.columns:
    df['cap-surface'] = df['cap-surface'].fillna(df['cap-surface'].mode()[0])

# Suppression des lignes avec trop de valeurs manquantes si nécessaire
# df.dropna(inplace=True)

# 4. Traitement des valeurs dupliquées
print("Nombre de doublons :", df.duplicated().sum())
df.drop_duplicates(inplace=True)

# 5. Visualisation du diamètre du chapeau (cap-diameter)
plt.figure(figsize=(10, 5))
plt.boxplot(df["cap-diameter"].dropna())
plt.title("Boîte à moustaches du diamètre du chapeau (cap-diameter)")
plt.ylabel("Diamètre")
plt.show()

# Calcul des valeurs aberrantes (Outliers) pour le diamètre
Q1 = df["cap-diameter"].quantile(0.25)
Q3 = df["cap-diameter"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df["cap-diameter"] < lower_bound) | (df["cap-diameter"] > upper_bound)]
print(f"Nombre de valeurs aberrantes pour le diamètre : {len(outliers)}")

# 6. Distributions et relations
# Distribution du diamètre
plt.figure(figsize=(10, 5))
plt.hist(df["cap-diameter"].dropna(), bins=20, color='skyblue', edgecolor='black')
plt.title("Distribution du diamètre du chapeau")
plt.xlabel("Diamètre")
plt.ylabel("Fréquence")
plt.show()

# Répartition des classes (Toxique 'p' vs Comestible 'e')
plt.figure(figsize=(6, 4))
df["class"].value_counts().plot(kind="bar", color=['red', 'green'])
plt.title("Distribution des classes (p = Toxique, e = Comestible)")
plt.xlabel("Classe")
plt.ylabel("Nombre")
plt.show()

# Relation entre hauteur du pied (stem-height) et largeur du pied (stem-width)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x="stem-height", y="stem-width", hue="class", alpha=0.5)
plt.title("Relation entre la hauteur et la largeur du pied")
plt.xlabel("Hauteur du pied")
plt.ylabel("Largeur du pied")
plt.show()
```
